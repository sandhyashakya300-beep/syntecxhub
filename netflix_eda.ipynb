```python
import numpy as np 
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
df['release_year'] = df['release_year'].fillna(0).astype('int')
df['duration'] = df['duration'].fillna('Unknown')
df['rating'] = df['rating'].fillna('Unknown')

```

```python
df.head(10  )
```

```python
type = df['type'].value_counts()
type
```

```python
plt.figure()
sns.barplot(x=type.index, y=type.values,palette='Set2')
plt.title('Distribution of Movies and TV Shows on Netflix', fontsize=10)
plt.xlabel('Type', )
plt.ylabel('Count', )
plt.savefig('netflix_type_distribution.png')
plt.show()
```

```python
print(df.columns.tolist())
```

```python
genres_series = df['genres'].str.split(', ').explode()
top_genres = genres_series.value_counts().head(10)
top_genres
```

```python
plt.figure()
sns.barplot(x=top_genres.values, y=top_genres.index,palette='viridis')
plt.title('Top 10 Genres on Netflix', fontsize=10,fontweight='bold')
plt.xlabel('Count', )
plt.ylabel('Genre', )
plt.savefig('netflix_top_genres.png', bbox_inches='tight')
plt.show()
```

```python
recent_growth = df[df['release_year'] >= 2015]
recent_growth_count = recent_growth.groupby(['release_year', 'type']).size().unstack(fill_value=0)
```

```python
plt.figure()
plt.plot(recent_growth_count.index, recent_growth_count['Movie'], marker='o', label='Movies')
plt.plot(recent_growth_count.index, recent_growth_count['TV Show'], marker='o', label='TV Shows')
plt.title('Growth of Movies and TV Shows on Netflix (2015-2024)', fontsize=10,fontweight='bold')
plt.xlabel('Release Year', )
plt.ylabel('Count', )
plt.legend()
plt.grid(True, linestyle='--', alpha=0.7)
plt.savefig('netflix_growth_trends.png', bbox_inches='tight')
plt.show()
```

```python
movies_df = df[df['type'] == 'Movie'].copy()
movies_df['duration_minutes'] = movies_df['duration'].astype(str).str.replace('(min)', '').astype('float')
plt.figure()
sns.histplot(movies_df['duration_minutes'], bins=30, kde=True, color='coral')
plt.title('Distribution of Movie Durations on Netflix', fontsize=10,fontweight='bold')  
plt.xlabel('Duration (minutes)' )
plt.ylabel('Frequency' )
plt.savefig('netflix_movie_durations.png', bbox_inches='tight')
plt.show()
```

```python
top_years = df['release_year'].value_counts().head(10).sort_index()
top_years
top_genres
```

```python
with open('netflix_eda_summary.txt', 'w') as f:
    f.write("Netflix EDA Summary\n")
    f.write("===================\n\n")
    
    f.write("1. Distribution of Movies and TV Shows:\n")
    f.write(f"   - Movies: {type['Movie']}\n")
    f.write(f"   - TV Shows: {type['TV Show']}\n\n")
    
    f.write("2. Top 10 Genres on Netflix:\n")
    for genre, count in top_genres.items():
        f.write(f"   - {genre}: {count}\n")
    
    f.write("\n3. Growth of Movies and TV Shows (2015-2024):\n")
    for year in recent_growth_count.index:
        f.write(f"   - {year}: Movies={recent_growth_count.loc[year, 'Movie']}, TV Shows={recent_growth_count.loc[year, 'TV Show']}\n")
    
    f.write("\n4. Distribution of Movie Durations:\n")
    f.write(f"   - Average Duration: {movies_df['duration_minutes'].mean():.2f} minutes\n")
    f.write(f"   - Median Duration: {movies_df['duration_minutes'].median():.2f} minutes\n")
```
