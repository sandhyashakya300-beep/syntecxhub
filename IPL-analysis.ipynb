```python
import pandas as pd
from matplotlib import pyplot as plt
import seaborn as sns
import numpy as np
```

```python
df = pd.read_csv('matches.csv')
df
```

```python
df.head(8)
```

```python
df.shape
```

```python
df_matches = df['season'].value_counts()
df_matches
```

```python
top_teams = df['winner'].value_counts().head(10)
top_teams
```

```python
top_scorer = df['player_of_match'].value_counts().head(10)
top_scorer
```

```python
df.columns.tolist()
```

```python
total_matches = df['team1'].value_counts() + df['team2'].value_counts()
total_matches
```

```python
total_wins = df['winner'].value_counts()
win_percentage = (total_wins / total_matches) * 100
win_percentage.sort_values(ascending=False, inplace=True)
win_percentage
```

```python
df.info()
```

```python
plt.figure()
sns.countplot(x='season', data=df, color='orange')
plt.xlabel('season')
plt.ylabel('number of matches')
plt.title('number of matches per season')
plt.tight_layout()
plt.show()
```

```python
plt.figure()
clean_winners = df[df['winner']!= 'no result']
winner_counts = clean_winners['winner'].value_counts()
sns.barplot(x=winner_counts.values, y=winner_counts.index, color='green')
plt.xlabel('number of matches won')
plt.ylabel('team')
plt.title('number of matches won by each team')
plt.tight_layout()
plt.show()
```

```python
plt.figure()
top_players = (df['player_of_match'].value_counts().iloc[:10])
sns.barplot(x=top_players.values, y=top_players.index, color='violet',hue=top_players.index,legend=False)
plt.xlabel('number of awards won')
plt.ylabel('player')
plt.title('top 10 players with most player of the match awards')
plt.tight_layout()
plt.show()
```

```python
plt.figure()
toss_decision_counts = df['toss_decision'].value_counts()
sns.barplot(x=toss_decision_counts.index, y=toss_decision_counts.values, color='cyan')
plt.xlabel('toss decision') 
plt.ylabel('number of matches')
plt.title('toss decisions in IPL matches')
plt.tight_layout()
plt.show()
```

```python
df['toss_win_match_win'] = np.where(df['toss_winner'] == df['winner'],'yes','no')
plt.figure()
sns.countplot(data=df[df['winner'] != 'no result'],x='toss_win_match_win',palette='Set2',hue='toss_win_match_win',legend=False,)
plt.title('Does winning the toss lead to winning the match?')
plt.xlabel('won both toss and matches?')
plt.ylabel('number of matches')
plt.show()
```
